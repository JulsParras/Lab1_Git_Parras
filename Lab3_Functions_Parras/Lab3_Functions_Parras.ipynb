```python
# [CELL 4] PARAMETRIC PROCESSING

def user_summary(title, *scores, **info):
    print(f"=== {title} ===")

    total = sum(scores)
    average = total / len(scores)

    print("Scores:", scores)
    print("Total:", total)
    print("Average:", round(average,2))

    for k,v in info.items():
        print(f"{k}: {v}")

    return average

avg = user_summary(
    f"{LAST_NAME} Academic Report",
    SEED_DIGIT * 10,
    ID_SUM % 100,
    NAME_LENGHT * 5,
    id=STUDENT_ID,
    surname=LAST_NAME
)
```

```python
# [CELL 5] RETURN MECHANISM

def compute_area(radius):
    return 3.1416 * radius **2

area = compute_area(SEED_DIGIT + 2)
print('Computed Area:', round(area,2))    
```

```python
# [CELL 6] SAFE CALCULATION

def safe_calculator(x,y):
    try:
        result = x/y
    except ZeroDivisionError:
        result = "UNDEFINED"
    return result

print('Calculation Result:', safe_calculator(ID_SUM, SEED_DIGIT))        
```

```python
# [CELL 7] RECURSION

def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n-1)

print('Factorial Result:', factorial(SEED_DIGIT))  
```

```python
# [CELL 8] STRING RECURSION

def reverse(text):
    if len(text) <= 1:
        return text
    return text[-1] + reverse(text[:-1])

print('Reversed Surname:', reverse(LAST_NAME.upper()))
```

```python
# [CELL 9] DECORATOR

def logger(func):
    def wrapper(*args, **kwargs):
        print('Executing:', func.__name__)
    return wrapper

@logger
def multiply(x,y):
    return x*y
print('Multiplication:', multiply(SEED_DIGIT, NAME_LENGHT))
```

```python
# [CELL 10] LAMBDA

data = list (range(1, SEED_DIGIT + 10))
squared = list(map(lambda x: x**2, data))

print('Squared Data:', squared)
```

```python
# [CELL 11] GENERATOR

def even_squares(limit):
    for i in range(limit):
        if i % 2 == 0:
            yield i**2

for value in even_squares (NAME_LENGHT + SEED_DIGIT):
    print(value)
    
```

```python
def compute_average(scores):
    return sum(scores) / len(scores)
def assign_grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C" 
    elif avg >= 60:
        return "D"
    else: 
        return "F"
def generate_remark(grade):
    remarks = {
        'A': 'Excellent Performance',
        'B': 'Good Performance',
        'C': 'Satisfactory Performance',
        'D': 'Needs Improvement',
        'F': 'Failing Status'
        }

    return remarks.get(grade, "Invalid Grade")
```

```python
 # main.py
import grades

# Ste\udent Identity Configuration
LAST_NAME = 'Parras' #Replace with your surname
STUDENT_ID = 'TUPM-26-3597' #Replace with your ID

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGHT = len(LAST_NAME)

# Generate student-unique scores
scores = [SEED_DIGIT * 10,
          ID_SUM % 100,
          NAME_LENGHT *7
          ]

average = grades.compute_average(scores)
grade = grades.assign_grade(average)
remark = grades.generate_remark(grade)

print ('=' * 40)
print(f'Student ID: {STUDENT_ID}')
print(f'Generated Scores: {scores}')
print(f'Average: {round(average,2)}')
print(f'Grade: {grade}')
print(f'Remark: {remark}')
print('='* 40)

```
